# **6. Crime counts and population**

For 2008 only, compare districts using the number of recorded incidents per 1,000 inhabitants:

```python
df_2008 = df_Chicago[df_Chicago["Year"] == 2008]
```

```python
df_2008.head()
```

```python
incidents_2008 = (df_2008.groupby("District").size()
    .reset_index(name="incidents"))
incidents_2008.head()
```

```python
#I used AI tools to organize and correct my error
population = pd.DataFrame({
    "District": range(1, 26),
    "Name": [
        "Central", "Wentworth", "Grand Crossing", "South Chicago",
        "Calumet", "Gresham", "Englewood", "Chicago Lawn",
        "Deering", "Ogden", "Harrison", "Monroe", "Wood",
        "Shakespeare", "Austin", "Jefferson Park", "Albany Park",
        "Near North", "Belmont", "Lincoln", "Prairie",
        "Morgan Park", "Town Hall", "Rogers Park", "Grand Central"
    ],
    "Population": [
        25613, 50957, 93384, 141422, 92729,
        105360, 91600, 244470, 165457, 137120,
        82392, 69677, 60517, 132459, 72736,
        199898, 156859, 110995, 107516, 102512,
        78111, 111545, 98391, 151435, 212535
    ]
})
population.head()
```

```python
#Merge the data
district_data = pd.merge( population, incidents_2008,
    on="District", how="left")  # keep all districts,
district_data
```

```python
district_data["incidents"] = district_data["incidents"].fillna(0).astype(int)
district_data["incidents_per_1000"]= (district_data["incidents"] / district_data["Population"] * 1000)
#district_data_s sorted
district_data_s = district_data.sort_values("incidents_per_1000", ascending=False)
print(district_data_s)
```

```python
# produced sorted bar chart.
plt.figure(figsize=(10, 7))

plt.barh(district_data["Name"], district_data["incidents_per_1000"])
plt.xlabel("Recorded incidents per 1,000 residents (2008)")
plt.title("Crime rate by district in 2008")
plt.gca().invert_yaxis()
plt.show()
```

**Interpretation:**
This bar chart shows a clear differences of crime rate across Chicago districts in 2008.

Highest rates is scored by **central** with 118. This is followed by **Wentworth** (67.64) and **Harrison** (59.9).  

**Lincon** scored the lowest rate with incidence rate of 12.7. Additionaly there is no record shown on  **Wood,Prairie and Town Hall** after filtering data set of 2008 there is **no** incidence record in this 3 districts.

# **7. Arrest proportion by crime type**

**Question:** count the number of incidents and compute the proportion with Arrest == True

```python
#I use AI tool on this part
# Group by crime type and calculate the count and the proportion of Arrest == True
arrest_stats = df_Chicago.groupby("Primary.Type").agg(
    incidents=("ID", "count"),                             # Step 1: count incidents
    arrest_proportion=("Arrest", lambda x: (x == True).mean()) # Step 2: proportion of True
).reset_index()
arrest_stats.head()
```

**Question:** keep only crime types with at least 100 recorded incidents;

```python
arrest_stats = arrest_stats[arrest_stats["incidents"] >= 100]
arrest_stats.head()
```

**Question:** sort the result by arrest proportion.

```python
# arrest proportion & sort
arrest_stats = arrest_stats.sort_values("arrest_proportion", ascending=False)
print(arrest_stats)
```

**Answer:** Result in horizontal bar chart


```python
plt.figure(figsize=(12, 6))
#horizontal chart
plt.barh(arrest_stats["Primary.Type"],arrest_stats["arrest_proportion"])

plt.xlabel("Proportion of incidents resulting in arrest")
plt.ylabel("Crime Type")
plt.title("Arrest Proportion by Crime Type")
plt.tight_layout()
plt.show()
```

#**Q 1:** Analyse monthly GAMBLING incidens


```python
gambling_df = df_Chicago[df_Chicago["Primary.Type"] == "GAMBLING"]

# Group by month and count
monthly_counts = gambling_df.set_index("Date").resample("ME").size()

plt.figure(figsize=(12, 4))
monthly_counts.plot(marker="o", color="blue")
plt.title("Monthly GAMBLING Incidents")
plt.xlabel("Month")
plt.ylabel("Number of Incidents")
plt.grid(True, alpha=0.3)
plt.show()
```

**Interpretation:** The chart show monthly gambling incidents in Chigago. In each year gambling is mosly recorded in month of July. The line also illustrates crime is decreasing from year to year, especialy there is big difference between 2008 and 2011.

#**Q 2**: At what time gambling recorded in the day?

```python
# Count incidents by hour
hour_counts = gambling_df["hour"].value_counts().sort_index()

# the chart
plt.figure(figsize=(10, 4))
plt.bar(hour_counts.index, hour_counts.values, color="darkgreen")
plt.title("GAMBLING Incidents by Hour of Day")
plt.xlabel("Hour (0 = Midnight)")
plt.ylabel("Number of Incidents")
plt.xticks(range(0, 24))
plt.show()
```

 **interpretation:** The gambling is mostly recorded between 18 hour to 20 hour. Additionaly the graph show medium frequency at 14 hour and 21 hour. However from midnight to 9:00 AM, it shows very low records.

#**Q3:** In which location gambling incidents mostly occur? Show with both table and a bar chart.

```python
# index top_locations
top_locations = gambling_df["location_clean"].value_counts().reset_index()

# Rename columns
top_locations.columns = ["Locations", "Number of Incidents"]
print(top_locations)
```
